```python
import re
import pandas as pd

# Extract likely measurement units from ingredient strings
standard_measure_pattern = re.compile(r"\b(?:grama|gramas|kg|g|ml|l|oz|lbs|mg|cm|mm|m)\b", re.IGNORECASE)
household_unit_pattern = re.compile(r"\b(?:xícara|xícaras|colher|colheres|copo|copos|pitada|pitadas|pacote|pacotes|envelope|envelopes|unidade|unidades|lata|latas|pedaço|pedaços|fatia|fatias|tbsp|tsp|cup|cups|caixa|caixas)\b(?:\s*\((?:chá|sopa)\))?", re.IGNORECASE)

units = []
for value in ingredients_df['ingredients'].dropna():
    units.extend(standard_measure_pattern.findall(str(value)))

sorted_units = sorted({u.lower() for u in units})
print(f'{sorted_units}')
print(f'{len(sorted_units)} unique standard measurement units found in the ingredients.')
# units
```

```python
import unicodedata

def convert_vulgar_fraction(text):
    text = list(text)
    fractions = [(i, unicodedata.normalize("NFKC", char)) for i, char in enumerate(text) if 'VULGAR FRACTION' in unicodedata.name(char)]
    if len(fractions) > 0:
        for pos, normalized_fraction in fractions:
            text[pos] = normalized_fraction
    return ''.join(text)

ing_a = '174 ml de água quente (⅔ de xícara (chá) + 1 colher (sopa))'
ing_a = convert_vulgar_fraction(ing_a)

ing_b = '⅔ de xícara (chá) de damasco seco cortado em cubos de 0,5 cm (cerca de 125 g)'
ing_b = convert_vulgar_fraction(ing_b)

ing_c = 'raspas de 30 limões sicilianos'
ing_c = convert_vulgar_fraction(ing_c)

ing_d = 'caldo de ½ limão'
ing_d = convert_vulgar_fraction(ing_d)

ing_e = '3 claras de ovo'
ing_e = convert_vulgar_fraction(ing_e)

ing_f = '½ pacote de gelatina em pó incolor e sem sabor (6 g)'
ing_f = convert_vulgar_fraction(ing_f)

ing_g = '2 ¼ latas de milho-verde (cerca de 2 xícaras [chá] de grãos escorridos)'
ing_g = convert_vulgar_fraction(ing_g)

entities = {'entities': []}

quantity_regex = '(?P<quantity>(\d+(?:\s*\d+)?(?:(?:\.|\,|\N{FRACTION SLASH})\d*)?))'
household_measures = 'xícaras|xícara|colheres|colher|copos|copo|pitadas|pitada|pacotes|pacote|envelopes|envelope|unidades|unidade|latas|lata|pedaços|pedaço|fatias|fatia|tbsp|tsp|cups|cup|caixas|caixa'
household_measure_regex = fr'(?P<mesure>(?:{household_measures})((?:\s*\((?:chá|sopa)\))|(?:\s*\[(?:chá|sopa)\]))?)'
standard_measures = 'grama|gramas|kg|g|ml|l|oz|lbs|mg'
standard_measure_regex = fr'(?P<mesure>{standard_measures})'

standard_measure_pattern = re.compile(fr'\b{quantity_regex}(?:\s*){standard_measure_regex}\b', re.IGNORECASE)
standard_match = standard_measure_pattern.search(ing_c)
if standard_match:
    entities['entities'].append((standard_match.start('quantity'), standard_match.end('mesure'), 'MEDIDA_PADRAO'))
    print(f"Standard measurement unit found: {standard_match.start('quantity')} - {standard_match.end('mesure')}: {standard_match.groupdict()} {standard_match.groups()})")

household_measure_pattern = re.compile(
    fr'\b{quantity_regex}(?:\s*\w*\s*){household_measure_regex}\s*(?:\+\s*)?\b',
    flags=re.IGNORECASE
)
household_match = household_measure_pattern.search(ing_c)
if household_match:
    entities['entities'].append((household_match.start('quantity'), household_match.end('mesure'), 'MEDIDA_CASEIRA'))
    print(f"Household measurement unit found: {household_match.start('quantity')} - {household_match.end('mesure')}: {household_match.groupdict()} {household_match.groups()}")

more_household_patterns = re.compile(
    fr'(?:\s*\+\s*{quantity_regex}(?:\s*\w*\s*){household_measure_regex})+', #mais recursão
    flags=re.IGNORECASE
)
more_household_matches = more_household_patterns.search(ing_c)
if more_household_matches:
    entities['entities'].append((more_household_matches.start('quantity'), more_household_matches.end('mesure'), 'MEDIDA_CASEIRA'))
    print(f"More household measurement unit found: {more_household_matches.start('quantity')} - {more_household_matches.end('mesure')}: {more_household_matches.groupdict()} {more_household_matches.groups()}")

# unit_regex = f'\s*(?P<quantity>\d+\s+\d+(?:\N{FRACTION SLASH}\d+)?|\d+(?:\N{FRACTION SLASH}\d+)?)\s+'
unit_regex = f'\s*(?P<quantity>\d+(?:\N{FRACTION SLASH}\d+)?)\s+(?!(?:de\s+)?(?:{household_measures}|{standard_measures}|cm|mm|m)(?:\s+|\)))'
unit_pattern = re.compile(fr'{unit_regex}', re.IGNORECASE)
unit_match = unit_pattern.search(ing_g)
if unit_match:
    entities['entities'].append((unit_match.start('quantity'), unit_match.end('quantity'), 'UNIDADE'))
    print(f"Quantity found: {unit_match.start('quantity')} - {unit_match.end('quantity')}: {unit_match.groupdict()} {unit_match.groups()}")

entities['entities'].sort(key=lambda x: x[0])
tupla = (ing_g, entities)
print(f"{tupla}")
```

```python
ingredients_df['ingredients_fractional'] = ingredients_df['ingredients'].apply(lambda x: convert_vulgar_fraction(x))

quantity_regex = '(?P<quantity>(\d+(?:\s*\d+)?(?:(?:\.|\,|\N{FRACTION SLASH})\d*)?))'
household_measures = 'xícaras|xícara|colheres|colher|copos|copo|pitadas|pitada|pacotes|pacote|envelopes|envelope|unidades|unidade|latas|lata|pedaços|pedaço|fatias|fatia|tbsp|tsp|cups|cup|caixas|caixa'
household_measure_regex = fr'(?P<mesure>(?:{household_measures})((?:\s*\((?:chá|sopa)\))|(?:\s*\[(?:chá|sopa)\]))?)'
standard_measures = 'grama|gramas|kg|g|ml|l|oz|lbs|mg'
standard_measure_regex = fr'(?P<mesure>{standard_measures})'
unit_regex = f'\s*(?P<quantity>\d+(?:\N{FRACTION SLASH}\d+)?\s+)(?!(?:de\s+)?(?:{household_measures}|{standard_measures}|cm|mm|m)(?:\s+|\)))'

standard_measure_pattern = re.compile(
    fr'\b{quantity_regex}(?:\s*){standard_measure_regex}\b', 
    flags=re.IGNORECASE
)
household_measure_pattern = re.compile(
    fr'\b{quantity_regex}(?:\s*\w*\s*){household_measure_regex}\s*(?:\+\s*)?\b',
    flags=re.IGNORECASE
)
more_household_patterns = re.compile(
    fr'(?:\s*\+\s*{quantity_regex}(?:\s*\w*\s*){household_measure_regex})+',
    flags=re.IGNORECASE
)
unit_pattern = re.compile(fr'{unit_regex}', re.IGNORECASE)

def create_entities(text):
    entities = {'entities': []}

    standard_match = standard_measure_pattern.search(text)
    if standard_match:
        entities['entities'].append((standard_match.start('quantity'), standard_match.end('mesure'), 'MEDIDA_PADRAO'))

    household_match = household_measure_pattern.search(text)
    if household_match:
        entities['entities'].append((household_match.start('quantity'), household_match.end('mesure'), 'MEDIDA_CASEIRA'))

    more_household_matches = more_household_patterns.search(text)
    if more_household_matches:
        entities['entities'].append((more_household_matches.start('quantity'), more_household_matches.end('mesure'), 'MEDIDA_CASEIRA'))

    unit_match = unit_pattern.search(text)
    if unit_match:
        entities['entities'].append((unit_match.start('quantity'), unit_match.end('quantity'), 'UNIDADE'))

    entities['entities'].sort(key=lambda x: x[0])
    return entities

ingredients_df['entities'] = ingredients_df['ingredients_fractional'].apply(create_entities)
ingredients_df
```
