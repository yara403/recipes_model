```python
import spacy
from pathlib import Path
import recipes_ner_model

outsiders_ingredients = [
    "1 kg de batata asterix (cerca de 6 unidades)",
    "1 1⁄4 de xícara (chá) de queijo parmesão ralado fino",
    "1 louro",
    "noz-moscada ralada na hora a gosto",
    "sal a gosto",
    "500 g de patinho moído",
    "1 cebola", 
    "2 dentes de alho", 
    "3 filés de anchova", 
    "1⁄2 xícara (chá) de ervilha congelada (cerca de 80 g)", 
    "2 colheres (sopa) de extrato de tomate", 
    "2 xícaras (chá) da água do cozimento das batatas", 
    "1 colher (sopa) de shoyu (molho de soja)",
    "1 colher (sopa) de vinagre de vinho tinto",
    "2 ramos de tomilho",
    "1 pitada de cravo em pó",
    "azeite a gosto",
    "sal e pimenta-do-reino moída na hora a gosto",
    "queijo parmesão ralado fino a gosto, para polvilhar"
]

model_path = Path(recipes_ner_model.PATH_TO_MODEL)
if not model_path.exists():
    recipes_ner_model.main()

recipes_ner_model = spacy.load(model_path)
for ingredient in outsiders_ingredients:
    doc = recipes_ner_model(ingredient)
    print(" ".join([f"{ent.text} ({ent.label_})" for ent in doc.ents]))
```

```python
text = "\n".join(ingredient for ingredient in outsiders_ingredients)
colors = {
    'MEDIDA_PADRAO': "#F16BE0", 
    'MEDIDA_CASEIRA': "#A6CAF3", 
    'MEDIDA_INFORMAL': '#98FB98', 
    'UNIDADE': "#FAC9D0", 
    'ALIMENTO': "#FCEA85"
    }
labels = ['MEDIDA_PADRAO', 'MEDIDA_CASEIRA', 'MEDIDA_INFORMAL', 'UNIDADE', 'ALIMENTO']
options = {'ents': labels, 'colors': colors}
spacy.displacy.render(recipes_ner_model(text), style='ent', jupyter=True, options=options)
```
